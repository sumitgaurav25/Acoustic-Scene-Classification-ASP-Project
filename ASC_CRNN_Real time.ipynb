```python
import numpy as np
import librosa
import tkinter as tk
from tensorflow.keras.models import load_model
from sklearn.preprocessing import LabelEncoder
import sounddevice as sd

# Load model and label encoder
model = load_model('E:/Users/Sumit/Downloads/UrbanSound8K/UrbanSound8K/saved_models_CRNN/audio_classification_CRNN.keras')
labelencoder = LabelEncoder()  # Assuming you have already fitted the encoder

# Function to classify real-time audio chunks
def classify_audio(audio_data, model, labelencoder):
    mfccs_features = librosa.feature.mfcc(y=audio_data, sr=44100, n_mfcc=40)
    mfccs_scaled_features = np.mean(mfccs_features.T, axis=0)
    prediction_feature = mfccs_scaled_features.reshape(1, 1, mfccs_scaled_features.shape[0], 1)
    predictions = model.predict(prediction_feature)
    predicted_class = np.argmax(predictions, axis=-1).flatten()
    return labelencoder.inverse_transform(predicted_class)[0]

# Replace this with the actual list of classes
class_labels = ["air_conditioner", "car_horn", "children_playing", "dog_bark", "drilling", "engine_idling", 
                "gun_shot", "jackhammer", "siren", "street_music"]  # Example UrbanSound8K classes

labelencoder = LabelEncoder()
labelencoder.fit(class_labels)

import pickle

# Save fitted label encoder
with open("label_encoder.pkl", "wb") as f:
    pickle.dump(labelencoder, f)

# Load the label encoder when needed
with open("label_encoder.pkl", "rb") as f:
    labelencoder = pickle.load(f)


# Set up GUI
root = tk.Tk()
root.title("Real-Time Acoustic Scene Classification")
label = tk.Label(root, text="Listening...", font=("Helvetica", 16))
label.pack()

# Update GUI label
def update_label(prediction):
    label.config(text=f"Detected Sound: {prediction}")

# Real-time audio callback
def audio_callback(indata, frames, time, status):
    audio_data = np.squeeze(indata)
    prediction = classify_audio(audio_data, model, labelencoder)
    update_label(prediction)

# Start audio stream and GUI loop
stream = sd.InputStream(callback=audio_callback, channels=1, samplerate=44100, blocksize=44100)
with stream:
    root.mainloop()
```
